DATA PREPROCESSING

```python
#writing the fuction to remove the punctuations
import string
def remove_punctuation(text):
  no_punctuation = "".join([i for i in text if i not in string.punctuation])
  return no_punctuation
```

```python
#writing the fucntion to tokenize
import nltk
nltk.download('punkt')
def tokenize(text):
  words = nltk.word_tokenize(text)
  return words
```

```python
#writing the function to remove stopwords
nltk.download('stopwords')
stopwords = nltk.corpus.stopwords.words('english')
def remove_stopwords(text):
  no_stopwords = [i for i in text if i not in stopwords]
  return no_stopwords
```

```python
#writing the function to lemmatize
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
def lemmatize(text):
  lemmatized_text = [lemmatizer.lemmatize(word) for word in text]
  return lemmatized_text
```

```python
#writing the function to do all the above functions at one place
def preprocess(df_col):
  corpus =[]
  for item in df_col:
    new_item = remove_punctuation(item)
    new_item = new_item.lower()
    new_item = tokenize(new_item)
    new_item = remove_stopwords(new_item)
    new_item = lemmatize(new_item)
    corpus.append(' '.join(str(x) for x in new_item))
  return corpus
```

```python
#saving the output into corpus
corpus = preprocess(data_reduced.review)
```

VECTORIZATION

```python
#using the bag of words to vectorize the text data
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(ngram_range=(1,2))
traindata = cv.fit_transform(corpus)
X = traindata
y = data_reduced.sentiment
```

```python
X.dtype
```

```python
#using random forest for classification
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(n_estimators=100)
clf.fit(X,y)
```

```python
#predicting the outputs using the model
from sklearn.metrics import accuracy_score
y_pred = clf.predict(X)
accuracy_score(y, y_pred)
```

```python
y.dtype
```

```python
#writing a function to classify the input text sentiment analysis
def find_sentiment(input_text):
  input_vector = cv.transform(preprocess(input_text))
  prediction = clf.predict(input_vector)
  if prediction == 0:
    print('Input statement has negative sentiment')
  if prediction == 1:
    print('Input statement has Positive sentiment')
```
